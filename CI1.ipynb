```python
# Define Fuzzy Sets
A = {'x1': 0.2, 'x2': 0.5, 'x3': 0.8}
B = {'x1': 0.6, 'x2': 0.3, 'x3': 0.4}
```

```python
# Union (max)
union = {x: max(A[x], B[x]) for x in A}
print("Union:", union)
```

```python
# Intersection (min)
intersection = {x: min(A[x], B[x]) for x in A}
print("Intersection:", intersection)
```

```python
# Complement (1 - μ)
comp_A = {x: 1 - A[x] for x in A}
print("Complement of A:", comp_A)
```

```python
# Difference (A - B = min(A, 1-B))
diff = {x: min(A[x], 1 - B[x]) for x in A}
print("Difference A-B:", diff)
```

```python
# Cartesian Product (Fuzzy Relation)
R = {}
for x in A:
    for y in B:
        R[(x, y)] = min(A[x], B[y])
print("Relation R:", R)
```

```python
# Another Relation (for composition)
C = {'y1': 0.7, 'y2': 0.5}
S = {}
for y in B:
    for z in C:
        S[(y, z)] = min(B[y], C[z])
print("Relation S:", S)
```

```python
# Max-Min Composition
T = {}
for x in A:
    for z in C:
        vals = []
        for y in B:
            vals.append(min(R[(x, y)], S[(y, z)]))
        T[(x, z)] = max(vals)

print("Max-Min Composition:", T)
```
